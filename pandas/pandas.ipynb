```python
arrays = [
   ["bar", "bar", "baz", "baz", "foo", "foo", "qux", "qux"],
   ["one", "two", "one", "two", "one", "two", "one", "two"],
]
```

```python
index = pd.MultiIndex.from_arrays(arrays, names=["first", "second"])
index
```

```python
df = pd.DataFrame(np.random.randn(8, 2), index=index, columns=["A", "B"])
df
```

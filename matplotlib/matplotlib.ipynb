# Pyplot API



```python
# Univariate: Numerical
data = {
    "Salary" : [25000, 30000, 37000, 28000, 39000, 48000, 55000, 52000, 35000, 26000, 27000, 31000, 33000, 29000, 40000, 36000, 34000, 45000, 47000, 50000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.shape
```

```python
# Line plot:
plt.plot(df['Salary'], color = 'purple', marker = "*", linestyle = "--", linewidth = 2)
plt.grid()
plt.show()
```

```python
# Histogram:
plt.hist(df["Salary"], bins = 5, color = "green")
plt.show()
```

```python
# Boxplot
plt.boxplot(df['Salary'])
plt.show()
```

```python
df.loc[20] = 0
```

```python
plt.boxplot(df['Salary'])
plt.show()
```

```python
df.drop(index = 20, inplace = True)
```

```python
df.shape
```

```python
# Univariate: Categorical
df["dept"] = ['HR', 'IT', 'Finance', 'HR', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR'] * 2
```

```python
df.head()
```

```python
# Pie Chart
count = df["dept"].value_counts()
count
```

```python
plt.pie(count, labels=count.index, autopct="%1.0f", explode=[0, 0.1, 0])
plt.axis("equal")
plt.show()
```

```python
# Countplot
plt.bar(count.index, count, color=["red", "green", "blue"])
plt.show()
```

```python
# Bivariate: Numerical, Numberical
df["Age"] = [22, 25, 29, 24, 30, 35, 40, 36, 28, 22, 23, 25, 26, 24, 31, 29, 27, 33, 35, 38]
df.head()
```

```python
# Scatter Plot
plt.scatter(df['Age'], df['Salary'], color='red')
plt.show()
```

```python
sort_age = df.sort_values("Age")
```

```python
# Line Plot
plt.plot(sort_age["Age"], df['Salary'])
plt.show()
```

```python
# Bar Chart
plt.bar(df['Age'], df['Salary'], color='Green')
plt.show()
```

```python
# Bivariate: Numerical, Categorical
```

```python
hr_sal = df[df["dept"] == "HR"]["Salary"]
hr_sal
```

```python
it_sal = df[df["dept"] == "IT"]["Salary"]
it_sal
```

```python
fin_sal = df[df["dept"] == "Finance"]["Salary"]
fin_sal
```

```python
# Box Plot
plt.boxplot([hr_sal, it_sal, fin_sal], labels=["HR", "IT", "FINANCE"])
plt.legend()
plt.show()
```

```python
# Pie Chart
salary_by_dept = df.groupby("dept")["Salary"].sum()
salary_by_dept
```

```python
plt.pie(salary_by_dept, labels=salary_by_dept.index, autopct="%1.2f")
plt.axis("equal")
plt.show()
```

```python
# bar plot
hr_mean = sum(hr_sal)/len(hr_sal)
it_mean = sum(it_sal)/len(it_sal)
fin_mean = sum(fin_sal)/len(fin_sal)
```

```python
plt.bar(["HR", "IT", "FINANCE"], [hr_mean, it_mean, fin_mean], color=["red", "blue", "green"])
# plt.grid()
plt.show()
```

```python
# Multivariate Analysis: 3 numerical column
df["experience"] =  [1, 2, 4, 1.5, 5, 8, 12, 10, 4, 1, 1.2, 2, 3, 1.8, 5.5, 4.5, 3.2, 7, 8, 10]
df.head()
```

```python
# Bubble plot
plt.scatter(df["Age"], df["Salary"], s = df["experience"]*10, color="red", edgecolors="black")
plt.title("Age vs Salary vs Experience")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
# Multivariate Analysis: 2 numerical and 1 categorical column
plt.scatter(df["Age"], df["Salary"], c=df["dept"].map({"HR": "yellow", "IT": "red", "Finance": "orange"}), edgecolors="black")
plt.title("Age vs Salary vs Department")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

```python
color = {"HR": "yellow", "IT": "blue", "Finance": "red"}
for dept, col in color.items():
  df_dept = df[df["dept"] == dept]
  plt.scatter(df_dept["Age"], df_dept["Salary"], c=col, label=dept, edgecolors="black")

plt.legend()
plt.title("Age vs Salary vs Department")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

# Object Oriented API

```python
fig, axs = plt.subplots(1, 3, figsize=(17 , 5))

# Line Plot
axs[0].plot(sort_age["Age"], df["Salary"], color="red")
axs[0].set_title("Line Plot")
axs[0].set_xlabel("Age")
axs[0].set_ylabel("Salary")

# Histogram
axs[1].hist(df["Salary"], bins=5, color="skyblue")
axs[1].set_title("Histogram")
axs[1].set_xlabel("Salary")
axs[1].set_ylabel("Frequency")

# Boxplot
axs[2].boxplot(df["Salary"])
axs[2].set_title("Boxplot")
axs[2].set_xlabel("Salary")

plt.savefig("multiple plots.jpg")
plt.show()
```

```python
data2 = {
    'Year': [2020, 2021, 2022, 2023],
    'Sales': [100, 150, 200, 250],
    'Profit': [20, 30, 40, 50],
    'Expenses': [80, 120, 160, 200]
}

df2 = pd.DataFrame(data2)
```

```python
plt.plot(df2["Year"], df2["Sales"], label = "Sales")
plt.plot(df2["Year"], df2["Profit"], label = "Profit")
plt.plot(df2["Year"], df2["Expenses"], label = "Expenses")

plt.title("Financial Analysis")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
# 3-D plot
ax = plt.axes(projection = "3d")

ax.scatter(df["Age"], df["Salary"], df["experience"])
ax.set_xlabel("Age")
ax.set_ylabel("Salary")
ax.set_zlabel("Experience")
plt.show()
```
